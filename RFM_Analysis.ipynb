```python
import pandas as pd

# 1. 讀取資料
df = pd.read_csv('RFM_raw.csv')
df['Date'] = pd.to_datetime(df['Date'].astype(str).str.split('.').str[0], format='%Y%m%d', errors='coerce')

# 2. 修改基準日期：把原本的 2019-01-13 往後調整 340 天
# 或者直接從 2019-01-04 (資料最大值) + 9天 = 2019-01-13
reference_date = pd.to_datetime('2019-01-13') 

# 3. 嘗試把 'count' 改回 'nunique'，如果數字還不對，請換成 'count'
result_df = df.groupby('CustomerKey').agg({
    'InvoNo': 'count',           # 如果這出來的 F 不是 25，請改成 'sum' 或其他邏輯
    'amount': 'sum',
    'Date': lambda x: (reference_date - x.max()).days
}).reset_index()

result_df.columns = ['Customer', 'F', 'M', 'R']
print(result_df.head())
```
